```python
import numpy as np, pandas as pd
rng = np.random.default_rng(0)
Xs = pd.DataFrame(rng.normal(size=(100, 5)), columns=["a", "b", "c", "d", "e"])
ys = 3 * Xs["a"] + 2 * Xs["c"] + rng.normal(size=100) * 0.5
```

```python
from sklearn.feature_selection import SelectKBest, f_regression

kb = SelectKBest(score_func=f_regression, k=4)
kb.fit(Xs, ys)
kb.transform(Xs)
```

```python
#help(SelectKBest)
print(kb.scores_.round(1))
print(kb.get_support())
print(kb.transform(Xs).shape)
```

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression

rfe = RFE(LinearRegression(), n_features_to_select=3)
rfe.fit(Xs,ys)
print(rfe.support_)
print(rfe.ranking_)
#help(RFE)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
pca.fit(Xs)
print(pca.explained_variance_ratio_)    

pca2 = PCA(n_components=0.95)
pca2.fit(Xs)
print(pca2.n_components_)
print(pca2.transform(Xs).shape)
```
