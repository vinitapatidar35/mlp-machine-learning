```python
from sklearn.model_selection import train_test_split
import numpy as np
df.loc[df["RM"]<=0,"RM"] = np.nan
df.loc[df["AGE"]<1,"AGE"] = np.nan
df.loc[df["RIVERSIDE"]=="UNKNOWN","RIVERSIDE"] =np.nan


X = df.drop(columns=["PRICE"])
y = df["PRICE"]
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25, random_state=0)
print(X_test.shape[0])
df.isna().sum()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import make_pipeline

ct = ColumnTransformer([
    ("crim", StandardScaler(),["CRIM"]),
    ("zn", MinMaxScaler(), ["ZN"] ),
    ("indus", MinMaxScaler(),["INDUS"]),
    ("polindex",StandardScaler(),["POLINDEX"]),
    ("dis",MinMaxScaler(),["DIS"]),
    ("highwaycount",MinMaxScaler(),["HIGHWAYCOUNT"]),
    ("tax",MinMaxScaler(),["TAX"]),
    ("ptratio",MinMaxScaler(),["PTRATIO"]),
    ("imm",MinMaxScaler(),["IMM"]),
    ("bpl",MinMaxScaler(),["BPL"]),
    ("rm",make_pipeline(SimpleImputer(strategy="mean"), MinMaxScaler()),["RM"]),
    ("age",make_pipeline(SimpleImputer(strategy="median"), StandardScaler()),["AGE"]),
    ("riverside",make_pipeline(SimpleImputer(strategy="most_frequent"), OneHotEncoder(drop='first',sparse_output=False)),["RIVERSIDE"]),
        
])

ct.fit(X_train)
ct.transform(X_train).shape[1]
```
